# Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
sns.set()
```

Load Data

```python
data = datasets.load_iris()
```

```python
data["data"][:5]
```

```python
data["feature_names"]
```

```python
data["target"]
```

Goal is to use attributes of the flower to predict the species of the flower. Specifically whether the flower is Setoa, Versicolor or Virginica. This will be done using the sepal length and width and petal length and width.

# Make Pandas Data Frame

```python
df = pd.DataFrame(data['data'], columns=data["feature_names"])
```

```python
df["target"] = data["target"]
```

# Basic Descriptive Stats

```python
df.describe()
```

# Distributions of Features and Target

```python
col = "sepal length (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

```python
col = "petal length (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

```python
col = "sepal width (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

```python
col = "petal width (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

# Relationship of features to Target

```python
df["target_name"] = df["target"].map({0:"setosa",1:"versicolor",2:"virginica"})
```

```python
col = "sepal length (cm)"
sns.relplot(x=col,y="target",hue="target_name", data=df)
plt.suptitle(col, y=1.01)
plt.show()
```

```python
col = "sepal width (cm)"
sns.relplot(x=col,y="target",hue="target_name", data=df)
plt.suptitle(col, y=1.01)
plt.show()
```

```python
col = "petal length (cm)"
sns.relplot(x=col,y="target",hue="target_name", data=df)
plt.suptitle(col, y=1.01)
plt.show()
```

```python
col = "petal width (cm)"
sns.relplot(x=col,y="target",hue="target_name", data=df)
plt.suptitle(col, y=1.01)
plt.show()
```

# EDA Pairplots

```python
sns.pairplot(df, hue="target_name")
```
